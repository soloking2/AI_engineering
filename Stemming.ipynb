```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
connect_tokens = ['connecting', 'connected', 'conectivity', 'connect', 'connects']
```

```python
for i in connect_tokens:
    print(i, ": ", ps.stem(i))
```

```python
learner_tokens = ['learning', 'learn', 'learned', 'learner', 'learns', 'leaners']
```

```python
for i in learner_tokens:
    print(i, ": ", ps.stem(i))
```

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
for i in learner_tokens:
    print(i, ": ", lemmatizer.lemmatize(i))
```

```python
likes_tokens = ['likes', 'better', 'worse']
```

```python
for i in likes_tokens:
    print(i, ": ", lemmatizer.lemmatize(i))
```
