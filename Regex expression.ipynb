```python
customer_reviews = ['great service from sara she found me what I wanted',
                    'lucy is such great addtion to the team',
                    'amazing work from sadeen',
                    'the cashier was very rude to me, I think her name was eleanor',
                    'sarah was able to find me find the items I needed quickly'
                   
                   ]
```

```python
sarahs_review = []
```

```python
pattern_to_find = r"sarah?"
```

```python
for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        sarahs_review.append(string)
```

```python
print(sarahs_review)
```

```python
a_reviews = []
pattern_to_find = r"^a" ##Starts with a

for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        a_reviews.append(string)
```

```python
print(a_reviews)
```

```python
y_review = []
pattern_to_find = r"y$" ##End starts with y

for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        y_review.append(string)
```

```python
print(y_review)
```

```python
needwant_reviews = []
pattern_to_find = r"(need|want)ed" ##Find sentences that contains needed or wanted

for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        needwant_reviews.append(string)
```

```python
print(needwant_reviews)
```

```python
no_punc_reviews = []
pattern_to_find = r"[^\w\s]"

for string in customer_reviews:
        no_punc_text = re.sub(pattern_to_find, "", string)
        no_punc_reviews.append(no_punc_text)
```

```python
print(no_punc_reviews)
```
